```python
#Tratamiento de Redundancias

df_urgencias.duplicated().sum()
```

```python
df_urgencias = df_urgencias.drop_duplicates()
```

```python
#Filtrado Booleano Simple

df_urgencias[
    df_urgencias["Edad_Paciente"] > 65
]
```

```python
#Filtrado Categórico por Lista (isin)

df_urgencias[
    df_urgencias["Hospital"].isin(
        ["Hospital Santa Sofía", "Clínica San Marcel"]
    )
]
```

```python
#Filtrado por Intervalo Clínico (between)

df_urgencias[
    df_urgencias["Tiempo_Espera_Min"].between(30.0, 90.0)
]
```

```python
#Consultas Directas con .query()

df_urgencias.query(
    "Nivel_Triage == 'Triage I' and Costo_Procedimiento > 300000"
)
```

```python
#Filtrado Compuesto con Métodos de Texto

df_urgencias[
    (df_urgencias["Destino_Alta"] == "Hospitalización") & (df_urgencias["Especialidad"].str.startswith("M"))
]
```

```python
#Imputación Estratificada de Faltantes: Rellena los valores ausentes (NaN) en la columna Tiempo_Espera_Min calculando su promedio general, y sustituye los nulos en Costo_Procedimiento por un costo base de 0.0 utilizando el método .fillna().

promedio_espera = df_urgencias["Tiempo_Espera_Min"].mean()
promedio_espera
```

```python
df_urgencias["Tiempo_Espera_Min"] = (
    df_urgencias["Tiempo_Espera_Min"].fillna(promedio_espera)
)
```

```python
df_urgencias["Tiempo_Espera_Min"].isna().sum()
```

```python
# sustituye los nulos en Costo_Procedimiento por un costo base de 0.0 utilizando el método .fillna().
df_urgencias["Costo_Procedimiento"] = (
    df_urgencias["Costo_Procedimiento"].fillna(0.0)
)

df_urgencias["Costo_Procedimiento"].isna().sum()
```

```python
#Transformación Personalizada de Atributos (apply)

def clasificar_edad(edad):
    if edad < 18:
        return "Pediatrico"
    elif edad <= 60:
        return "Adulto"
    else:
        return "Adulto Mayor"


df_urgencias["Grupo_Etario"] = (
    df_urgencias["Edad_Paciente"].apply(clasificar_edad)
)

df_urgencias[
    ["Edad_Paciente", "Grupo_Etario"]
].head()
```

```python
#Renombrado y Reorganización de Campos

df_urgencias = df_urgencias.rename(
    columns={
        "Tiempo_Espera_Min": "Minutos_Espera",
        "Destino_Alta": "Egreso"
    }
)

df_urgencias.columns
```

```python
# reordena las columnas colocando Grupo_Etario en la primera posición de la tabla.

df_urgencias.columns

columnas = ["Grupo_Etario"] + [
    col for col in df_urgencias.columns
    if col != "Grupo_Etario"
]

df_urgencias = df_urgencias[columnas]

df_urgencias.head()
```

```python
#Agregación por Centro Hospitalario

resumen_hospital = df_urgencias.groupby("Hospital").agg(
    Promedio_Espera=("Minutos_Espera", "mean"),
    Total_Costos=("Costo_Procedimiento", "sum")
)

resumen_hospital
```

```python
#Medición de Rango Clínico

def rango(series):
    return series.max() - series.min()

rango_triage = df_urgencias.groupby("Nivel_Triage").agg(
    Rango_Espera=("Minutos_Espera", rango),
    Rango_Costo=("Costo_Procedimiento", rango)
)

rango_triage
```

```python
#Agrupación Multinivel Avanzada

resumen_hospital_triage = df_urgencias.groupby(
    ["Hospital", "Nivel_Triage"]
).agg(
    Promedio_Edad=("Edad_Paciente", "mean"),
    Total_Costos=("Costo_Procedimiento", "sum")
)

resumen_hospital_triage
```
